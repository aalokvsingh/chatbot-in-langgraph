```python
@tool
def get_conversion_factor(base_currency: str, target_currency: str) -> float:
  """
  This function fetches the currency conversion factor between a given base currency and a target currency
  """
  url = f'https://v6.exchangerate-api.com/v6/c754eab14ffab33112e380ca/pair/{base_currency}/{target_currency}'

  response = requests.get(url)

  return response.json()
```

```python
from typing import Annotated
from langchain_core.tools import InjectedToolArg

@tool
def convert(base_currency_value: int, conversion_rate: Annotated[float, InjectedToolArg]) -> float:
  """
  given a currency conversion rate this function calculates the target currency value from a given base currency value
  """

  return base_currency_value * conversion_rate
```

```python
get_conversion_factor.invoke({"base_currency": "USD", "target_currency": "INR"})
```

```python
convert.invoke({"base_currency_value": 100, "conversion_rate": 82.5})  
```

```python
llm_with_tools = llm.bind_tools([get_conversion_factor, convert])   
```

```python
messages = [HumanMessage(content="What is the conversion factor for USD to INR and also convert 10 USD to INR?")]
messages
```

```python
ai_messages = llm_with_tools.invoke(messages)
messages.append(ai_messages)
```

```python
messages[1].tool_calls
```

```python
import json

for tool_call in ai_messages.tool_calls:
    if tool_call['name'] == "get_conversion_factor":
        conversion_factor_response = get_conversion_factor.invoke(tool_call)
        conversion_factor = json.loads(conversion_factor_response.content)['conversion_rate']
        # print(conversion_factor['conversion_rate'])
        messages.append(conversion_factor_response)

    elif tool_call['name'] == "convert":
        conversion_rate = conversion_factor
        convert_response = convert.invoke({"base_currency_value": tool_call['args']['base_currency_value'], "conversion_rate": conversion_rate})
        print(convert_response)
```
